Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]
count = 0

for user in users:
    if count == 3:
        print("Limit reached. Stop sending emails.")
        break
    print(f"Email sent to {user}")
    count += 1
```

Q3. Calculate total price of items in a shopping cart (Using For loop)

cart = [100, 250, 50, 300]

Output:

Total Price: 700

```python
cart = [100, 850, 60, 300]
total_price = 0

for price in cart:
    total_price += price

print(f"Total Price: {total_price}")
```

Q4. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True}
]

Output:

Access granted to: Ali

Access granted to: John

```python
customers = [
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "Jhon", "paid": True}
]

for customer in customers:
    if not customer["paid"]:
        continue
    print(f"Access granted to: {customer['name']}")
```

Q5. Send personalized emails to a list of users (Using for loop)

users = ["Ali", "Sara", "John"]

Output:

Hello Ali, your subscription is active!
Hello Sara, your subscription is active!
Hello John, your subscription is active!

Q6. Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']



```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domains = []

for email in emails:
    domain = email.split("@")[1]
    domains.append(domain)

print(f"domains: {domains}")
```

Q7. Create a list of numbers, generate a new list containing only the numbers that are divisible by 3 or 5. (List Comprehension)

```python

numbers = [1, 3, 5, 9, 10, 12, 14, 15, 20, 22]
result = [num for num in numbers if num % 3 == 0 or num % 5 == 0]

print(result)
```

Q8. Create a multiplication table using a nested for loop.

numbers = [2, 3, 4]
multipliers = [1, 2, 3]

Expected Output:

2 x 1 = 2
2 x 2 = 4
2 x 3 = 6
3 x 1 = 3
3 x 2 = 6
3 x 3 = 9
4 x 1 = 4
4 x 2 = 8
4 x 3 = 12

Use a nested for loop.

```python
numbers = [2, 3, 4]
multipliers = [1, 2, 3]

for num in numbers:
    for mult in multipliers:
        print(f"{num} x {mult} = {num * mult}", end=" ")
```

Q9. Keep asking the user for a password until the correct password is entered.

correct_password = "python123"

Expected behaviour:

Enter password: abc
Incorrect password. Try again.

Enter password: 12345
Incorrect password. Try again.

Enter password: python123
Access granted!

Use a while loop.

```python
correct_password = "123"

while True:
    password = input("Enter password: ")
    if password == correct_password:
        print("Access granted!")
        break
    else:
        print("Incorrect password. Try again.")
```
